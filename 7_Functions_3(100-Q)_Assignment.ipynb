```python
# Assignment 12: Check Leap Year
# Write a function named is_leap_year().

# Requirements
# Accept a year as an integer.
# Return True when it is a leap year.
# Otherwise, return False.
# Print an appropriate message.
# Rules
# A leap year is:

# Divisible by 400, or
# Divisible by 4 but not divisible by 100.

# Function to check whether a year is a leap year
def is_leap_year(year):
    # Check leap year conditions
    if year % 400 == 0 or (year % 4 == 0 and year % 100 != 0):
        return True
    else:
        return False

# Main program

# Accept year from the user
year = int(input("Enter a year: "))

# Call the function
result = is_leap_year(year)

# Display appropriate message
if result:
    print(year, "is a leap year.")
else:
    print(year, "is not a leap year.")
```

```python
# Assignment 13: Factorial Using a Function
# Write a function named calculate_factorial().

# Requirements
# Accept a non-negative integer.
# Calculate its factorial using a loop.
# Do not use math.factorial().
# Return the factorial.
# Handle negative input.

# Function to calculate factorial of a number
def calculate_factorial(number):
    # Handle negative input
    if number < 0:
        return "Error: Factorial is not defined for negative numbers."

    # Calculate factorial using a loop
    factorial = 1
    for i in range(1, number + 1):
        factorial = factorial * i
    return factorial

# Main program

# Accept a non-negative integer from the user
number = int(input("Enter a non-negative integer: "))

# Call the function
result = calculate_factorial(number)

# Display the result
print("Factorial:", result)
```

```python
# Assignment 16: Reverse a String
# Write a function named reverse_string().

# Requirements
# Accept a string.
# Return its reversed version.
# Solve it first using slicing.
# Write another function that reverses it using a loop.

# Function to reverse a string using slicing
def reverse_string(text):
    # Reverse the string using slicing
    return text[::-1]

# Function to reverse a string using a loop
def reverse_string_loop(text):
    # Initialize an empty string
    reversed_text = ""

    # Add characters in reverse order using a loop
    for char in text:
        reversed_text = char + reversed_text
    return reversed_text

# Main program

# Accept a string from the user
text = input("Enter a string: ")

# Call slicing function
result1 = reverse_string(text)

# Call loop function
result2 = reverse_string_loop(text)

# Display results
print("Reversed string using slicing:", result1)
print("Reversed string using loop:", result2)
```

```python
# Assignment 17: Palindrome Checker
# Write a function named is_palindrome().

# Requirements
# Accept a string.
# Ignore spaces and letter case.
# Return True when the string is a palindrome.
# Print a meaningful result.

# Function to check whether a string is a palindrome
def is_palindrome(text):
    # Remove spaces and convert string to lowercase
    cleaned_text = text.replace(" ", "").lower()

    # Check if the string is equal to its reverse
    if cleaned_text == cleaned_text[::-1]:
        return True
    else:
        return False

# Main program

# Accept a string from the user
text = input("Enter a string: ")

# Call the function
result = is_palindrome(text)

# Display meaningful result
if result:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

```python
# Assignment 19: Find Maximum in a List
# Write a function named find_list_maximum().

# Requirements
# Accept a list of numbers.
# Do not use max().
# Return the largest list element.
# Handle an empty list by returning an appropriate message.

# Function to find the maximum element in a list
def find_list_maximum(numbers):
    # Check if the list is empty
    if len(numbers) == 0:
        return "The list is empty."

    # Assume the first element is the largest
    largest = numbers[0]

    # Compare each element using a loop
    for number in numbers:
        if number > largest:
            largest = number
    return largest

# Main program

# Accept list elements from the user
numbers = []

count = int(input("Enter the number of elements: "))

for i in range(count):
    value = int(input("Enter number: "))
    numbers.append(value)

# Call the function
result = find_list_maximum(numbers)

# Display the result
print("Largest element in the list:", result)
```

```python
# Assignment 30: Mixed Parameter Types
# Write a function that contains:

# Positional-only parameters
# Normal parameters
# Default parameters
# Variable positional arguments
# Keyword-only parameters
# Variable keyword arguments
# Requirements
# Create a function named generate_report() and demonstrate all parameter types in one complete program.

# Hint
# Study this general order:

# def function(positional_only, /, normal, default=value, *args, keyword_only, **kwargs):


# Function demonstrating all types of parameters
def generate_report(report_id, /, student_name, course="Python", *marks, grade="A", **details):

    # Positional-only parameter
    print("Report ID:", report_id)

    # Normal parameter
    print("Student Name:", student_name)

    # Default parameter
    print("Course:", course)

    # Variable positional arguments
    print("Marks:")
    for mark in marks:
        print(mark)

    # Keyword-only parameter
    print("Grade:", grade)

    # Variable keyword arguments
    print("Additional Details:")
    for key, value in details.items():
        print(key, ":", value)

# Main program

# Calling function with all parameter types
generate_report(
    101,                    # Positional-only argument
    "Anil",                 # Normal argument
    "Python Programming",   # Default parameter value changed
    85, 90, 95,             # Variable positional arguments (*marks)
    grade="A+",             # Keyword-only argument
    city="Bengaluru",       # Variable keyword argument (**details)
    qualification="BCA"
)

# Calling function using default course value
generate_report(
    102,
    "John",
    grade="B",
    city="New York",
    qualification="MCA"
)
```

```python
# Assignment 31: Return Multiple Values
# Write a function named number_statistics().

# Requirements
# Accept a list of numbers.
# Return minimum, maximum, total, and average.
# Use tuple unpacking to store returned values.
# You may use built-in functions in this assignment.

# Function to calculate statistics of a list of numbers
def number_statistics(numbers):
    # Calculate minimum, maximum, total, and average
    minimum = min(numbers)
    maximum = max(numbers)
    total = sum(numbers)
    average = total / len(numbers)

    # Return multiple values as a tuple
    return minimum, maximum, total, average

# Main program

# Accept numbers from the user
numbers = []
count = int(input("Enter the number of elements: "))
for i in range(count):
    value = float(input("Enter number: "))
    numbers.append(value)

# Use tuple unpacking to store returned values
minimum, maximum, total, average = number_statistics(numbers)

# Display results
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Total:", total)
print("Average:", round(average, 2))
```

```python
# Assignment 32: Function Returning a Dictionary
# Write a function named analyze_text().

# Requirements
# Accept a string.
# Return a dictionary containing:
# Total characters
# Total words
# Total vowels
# Total digits
# Total spaces
# Display every result clearly.

# Function to analyze text and return details as a dictionary
def analyze_text(text):
    # Count total characters
    total_characters = len(text)

    # Count total words
    total_words = len(text.split())

    # Count vowels
    vowels = "aeiouAEIOU"
    total_vowels = 0

    for char in text:
        if char in vowels:
            total_vowels += 1

    # Count digits and spaces
    total_digits = 0
    total_spaces = 0

    for char in text:
        if char.isdigit():
            total_digits += 1
        elif char == " ":
            total_spaces += 1

    # Return results as a dictionary
    return {
        "Total Characters": total_characters,
        "Total Words": total_words,
        "Total Vowels": total_vowels,
        "Total Digits": total_digits,
        "Total Spaces": total_spaces
    }

# Main program

# Accept a string from the user
text = input("Enter a text: ")

# Call the function
result = analyze_text(text)

# Display every result clearly

for key, value in result.items():
    print(key, ":", value)
```

```python
# Assignment 33: Function Returning a List
# Write a function named get_even_numbers().

# Requirements
# Accept a starting number and ending number.
# Return a list containing all even numbers in the range.
# Validate that the starting number is smaller than the ending number.

# Function to get even numbers in a given range
def get_even_numbers(start, end):
    # Validate that starting number is smaller than ending number
    if start >= end:
        return "Invalid range: Starting number must be smaller than ending number."

    # Create a list to store even numbers
    even_numbers = []

    # Check each number in the range
    for number in range(start, end + 1):
        if number % 2 == 0:
            even_numbers.append(number)

    return even_numbers


# Main program

# Accept starting and ending numbers from the user
start = int(input("Enter starting number: "))
end = int(input("Enter ending number: "))

# Call the function
result = get_even_numbers(start, end)

# Display the result
print("Even numbers in the range:", result)
```

```python
# Assignment 34: Prime Number Function
# Write a function named is_prime().

# Requirements
# Accept one integer.
# Return True when the number is prime.
# Return False otherwise.
# Use the function to print all prime numbers from 1 to 100.

# Hint
# A prime number has exactly two factors.

# Function to check whether a number is prime
def is_prime(number):
    # Numbers less than 2 are not prime
    if number < 2:
        return False

    # Check for factors from 2 to number-1
    for i in range(2, number):
        if number % i == 0:
            return False
    return True

# Main program

# Print all prime numbers from 1 to 100
print("Prime numbers from 1 to 100:")

for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

```python
# Assignment 35: Fibonacci Function
# Write a function named generate_fibonacci().

# Requirements
# Accept the number of terms.
# Return the complete Fibonacci series as a list.
# Handle zero and negative term counts.
# Do not use recursion in this assignment.

# Function to generate Fibonacci series
def generate_fibonacci(terms):
    # Handle zero and negative term counts
    if terms <= 0:
        return "Number of terms must be greater than zero."

    # Initialize Fibonacci series
    fibonacci = []

    # First two Fibonacci numbers
    a = 0
    b = 1

    # Generate Fibonacci series using a loop
    for i in range(terms):
        fibonacci.append(a)
        a, b = b, a + b
    return fibonacci

# Main program

# Accept number of terms from the user
terms = int(input("Enter number of Fibonacci terms: "))

# Call the function
result = generate_fibonacci(terms)

# Display the result
print("Fibonacci series:", result)
```

```python
# Assignment 56: Mask Email Addresses
# Given a list of email addresses, mask the username portion.

# Example
# bharath@gmail.com → b******@gmail.com

# Requirements
# Use map().
# Keep the first character of the username.
# Replace remaining username characters with *.
# Keep the domain unchanged.
# Create a separate named function for masking.

# Function to mask email addresses
def mask_email(email):
    # Split email into username and domain
    username, domain = email.split("@")

    # Keep first character and replace remaining characters with *
    masked_username = username[0] + "*" * (len(username) - 1)

    # Return masked email
    return masked_username + "@" + domain

# Main program

# Accept email addresses from the user
emails = input("Enter email addresses: ").split()

# Use map() with the named function
masked_emails = list(map(mask_email, emails))

# Display original and masked emails
print("Original Emails:", emails)
print("Masked Emails:", masked_emails)
```

```python
# Assignment 67: Filter Prime Numbers
# Accept a list of integers.

# Requirements
# Write a reusable is_prime() function.
# Pass the function to filter().
# Print only prime values.
# Handle values below 2 correctly.

# Function to check whether a number is prime
def is_prime(n):
    # Numbers less than 2 are not prime
    if n < 2:
        return False

    # Check if n is divisible by any number from 2 to sqrt(n)
    for i in range(2, int(n ** 0.5) + 1):
        # If divisible, the number is not prime
        if n % i == 0:
            return False
    # If no divisor was found, the number is prime
    return True

# List of integers
numbers = [1,2,3,4,30,20,10,11,12,13,-2,-3,0]

# Use filter() with the is_prime function
prime_numbers = list(filter(is_prime, numbers))

# Print only the prime numbers
print(prime_numbers)
```

```python
# Assignment 68: Filter Palindrome Words
# Accept a sentence from the user.

# Requirements
# Split it into words.
# Use filter() to find palindrome words.
# Ignore case and punctuation.
# Write a separate function for palindrome checking.

# Function to check whether a word is a palindrome
def is_palindrome(word):
    # Convert the word to lowercase
    word = word.lower()

    # Remove punctuation and keep only letters and numbers
    palindrome = ""
    for ch in word:
        if ch.isalnum():
            palindrome += ch

    # Check if the palindrome is the same forwards and backwards
    return palindrome == palindrome[::-1]

# Accept a sentence from the user
sentence = input("Enter a sentence: ")

# Split the sentence into individual words
words = sentence.split()

# Use filter() to find palindrome words
palindrome_words = list(filter(is_palindrome, words))

# Print the palindrome words
print("Palindrome words:", palindrome_words)
```

```python
# Assignment 74: Filter Perfect Squares
# Accept a list of positive integers.

# Requirements
# Use filter() to return values that are perfect squares.
# Write a named function for checking perfect squares.
# Use the math module if required.
# Example
# Input: [1, 2, 4, 8, 9, 15, 16]
# Output: [1, 4, 9, 16]

import math

# Function to check whether a number is a perfect square
def is_perfect_square(n):
    # Find the square root of the number
    root = math.isqrt(n)

    # A number is a perfect square if its square root
    # multiplied by itself gives the original number
    return root ** 2 == n

# Accept a list of integers from the user
numbers = list(map(int, input("Enter numbers: ").split()))

# Use filter() with the named function
perfect_squares = list(filter(is_perfect_square, numbers))

# Print the perfect squares
print("Perfect squares:", perfect_squares)
```

```python
# Assignment 75: Filter Dates by Year
# Given dates in DD-MM-YYYY format:

# Requirements
# Ask the user for a year.
# Use filter() to return dates belonging to that year.
# Handle invalid date strings safely.
# Use datetime.strptime() for validation.

from datetime import datetime

# List of dates
dates = [
    "15-08-2024",
    "20-01-2025",
    "10-12-2024",
    "31-02-2024",
    "05-06-2025",
    "hello",
    "25-12-2024"
]

# Ask the user for a year
year = int(input("Enter year: "))

# Function to check the date
def is_date_in_year(date):
    # Check whether the date has the correct format
    parts = date.split("-")
    if len(parts) != 3:
        return False

    # Check whether day, month and year are numbers
    if not all(part.isdigit() for part in parts):
        return False

    day, month, date_year = map(int, parts)

    # Basic range checking before using strptime()
    if month < 1 or month > 12:
        return False

    if day < 1 or day > 31:
        return False

    # Check February
    if month == 2 and day > 29:
        return False

    # Months with only 30 days
    if month in [4, 6, 9, 11] and day > 30:
        return False

    # Validate the date using datetime.strptime()
    valid_date = datetime.strptime(date, "%d-%m-%Y")

    # Return True if the year matches
    return valid_date.year == year


# Use filter() to find dates from the selected year
filtered_dates = list(filter(is_date_in_year, dates))

# Print the result
print(filtered_dates)
```

```python
# Assignment 81: Calculate Factorial Using reduce()
# Write a function that calculates factorial using reduce().

# Requirements
# Accept a non-negative integer.
# Generate numbers from 1 to the given number.
# Multiply them using reduce().
# Return 1 when the input is 0.
# Reject negative values.

# Import reduce from functools
from functools import reduce

# Function to calculate factorial using reduce()
def factorial(n):

    # Reject negative values
    if n < 0:
        return "Negative values are not allowed."

    # Generate numbers from 1 to n
    numbers = range(1, n + 1)

    # Multiply all numbers using reduce()
    # Start with 1 so that factorial(0) returns 1
    result = reduce(lambda accumulator, value: accumulator * value, numbers, 1)
    return result

# Accept a non-negative integer from the user
number = int(input("Enter a non-negative integer: "))

# Calculate and print the factorial
print("Factorial:", factorial(number))
```

```python
# Assignment 91: Closure-Based Counter
# Write a function named create_counter().

# Requirements
# Define a variable count = 0 in the outer function.
# Define an inner function that increases the count.
# Use nonlocal to modify the enclosing variable.
# Return the inner function.
# Call the returned function at least five times.
# Expected Output
# 1
# 2
# 3
# 4
# 5

# Outer function
def create_counter():

    # Variable in the outer function
    count = 0

    # Inner function
    def counter():

        # Use nonlocal to modify the count
        # from the enclosing function
        nonlocal count

        # Increase count by 1
        count += 1

        # Return the updated count
        return count

    # Return the inner function
    return counter

# Create a counter
counter = create_counter()

# Call the returned function five times
print(counter())
print(counter())
print(counter())
print(counter())
print(counter())
```

```python
# Assignment 92: Closure-Based Bank Balance
# Write a function named create_bank_account().

# Requirements
# Store the balance in the outer function.
# Define inner functions for:
# Deposit
# Withdrawal
# Balance enquiry
# Use nonlocal to update the balance.
# Reject withdrawals greater than the available balance.
# Return all three inner functions.
# Hint
# Return multiple functions:

# return deposit, withdraw, check_balance


# Outer function to create a bank account
def create_bank_account(initial_balance):

    # Store the balance in the outer function
    balance = initial_balance

    # Inner function for depositing money
    def deposit(amount):
        nonlocal balance

        # Add the deposit amount to the balance
        balance += amount

        print("Deposited:", amount)

    # Inner function for withdrawing money
    def withdraw(amount):
        nonlocal balance

        # Check whether enough balance is available
        if amount > balance:
            print("Insufficient balance")
        else:
            # Subtract the withdrawal amount
            balance -= amount
            print("Withdrawn:", amount)

    # Inner function for checking the balance
    def check_balance():
        print("Balance:", balance)

    # Return all three inner functions
    return deposit, withdraw, check_balance

# Create a bank account with an initial balance of 1000
deposit, withdraw, check_balance = create_bank_account(1000)

# Check the initial balance
check_balance()

# Deposit money
deposit(500)

# Check the balance
check_balance()

# Withdraw money
withdraw(300)

# Check the balance
check_balance()

# Try to withdraw more than the available balance
withdraw(1500)

# Check the final balance
check_balance()
```

```python
# Assignment 93: Function Decorator
# Create a decorator named execution_logger().

# Requirements
# The decorator should print a message before the original function executes.
# It should print another message after execution.
# Apply it to an addition function.
# Preserve the original function's return value.
# Expected Pattern
# Starting function...
# Result: 30
# Function completed.

# Create the decorator
def execution_logger(func):

    # Wrapper function that runs before and after the original function
    def wrapper(a, b):

        # Message before the function executes
        print("Starting function...")

        # Call the original function and store its return value
        result = func(a, b)

        # Print the result
        print("Result:", result)

        # Message after the function executes
        print("Function completed.")

        # Return the original function's result
        return result

    # Return the wrapper function
    return wrapper

# Apply the decorator to the addition function
@execution_logger
def addition(a, b):
    return a + b

# Call the decorated function
addition(10, 20)
```

```python
# Assignment 100: Employee Performance Analysis System
# ============================================================
# EMPLOYEE PERFORMANCE ANALYSIS SYSTEM
# ============================================================

# Import reduce from functools
from functools import reduce

# ============================================================
# SAMPLE EMPLOYEE DATA
# ============================================================

employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]

# ============================================================
# GLOBAL SCOPE
# ============================================================

# This variable belongs to the global scope.
company_name = "Edukron"

# ============================================================
# VALIDATE EMPLOYEE DATA
# ============================================================

def validate_employee(employee):
    """
    Validate salary and performance score.

    Salary must be zero or greater.
    Performance score must be between 0 and 100.
    """

    salary = employee["salary"]
    performance = employee["performance_score"]

    # Validate salary
    if not isinstance(salary, (int, float)) or salary < 0:
        return False

    # Validate performance score
    if not isinstance(performance, (int, float)):
        return False

    if performance < 0 or performance > 100:
        return False

    return True

# ============================================================
# PART 1: DISPLAY EMPLOYEES
# ============================================================

def display_employees(employee_list):
    """Display all employee details."""

    # Handle empty employee list
    if not employee_list:
        print("\nNo employees available.")
        return
    print("\n========== EMPLOYEE DETAILS ==========")

    for employee in employee_list:

        print("--------------------------------------")
        print("ID                :", employee["id"])
        print("Name              :", employee["name"])
        print("Department        :", employee["department"])
        print("Salary            :", employee["salary"])
        print("Performance Score :", employee["performance_score"])
        print("Experience        :", employee["experience"], "years")

    print("--------------------------------------")


# ============================================================
# PART 2: EXTRACT EMPLOYEE NAMES USING map()
# ============================================================

def get_employee_names(employee_list):
    """Return employee names using map()."""

    # map() applies the lambda to every employee.
    names = list(map(lambda employee: employee["name"], employee_list))

    return names


# ============================================================
# PART 3: FILTER HIGH PERFORMERS
# ============================================================

def get_high_performers(employee_list):
    """Return employees with performance score >= 80."""

    # filter() selects employees whose score is 80 or above.
    high_performers = list(filter(lambda employee: employee["performance_score"] >= 80,employee_list))

    return high_performers

# ============================================================
# PART 4: FILTER BY DEPARTMENT
# ============================================================

def get_by_department(employee_list, department):
    """Return employees belonging to a department."""

    # Convert the search to lowercase for case-insensitive matching.
    department = department.strip().lower()

    # filter() selects matching employees.
    result = list(filter(lambda employee:employee["department"].lower() == department, employee_list ))

    return result

# ============================================================
# PART 5: CALCULATE REVISED SALARIES
# ============================================================

def calculate_revised_salaries(employee_list):
    """
    Use map() to create new employee dictionaries
    with revised salaries.

    The original employee dictionaries are not modified.
    """

    def calculate_salary(employee):
        score = employee["performance_score"]
        salary = employee["salary"]

        # Decide the percentage increase.
        if score >= 90:
            increase = 0.15
        elif score >= 80:
            increase = 0.10
        elif score >= 70:
            increase = 0.05
        else:
            increase = 0

        # Calculate the new salary.
        revised_salary = salary + (salary * increase)

        # Create a copy instead of modifying the original dictionary.
        new_employee = employee.copy()

        # Store the revised salary in the copy.
        new_employee["revised_salary"] = revised_salary

        return new_employee

    # map() applies calculate_salary() to every employee.
    revised_employees = list(map(calculate_salary, employee_list))

    return revised_employees


# ============================================================
# PART 6: TOTAL SALARY EXPENSE
# ============================================================

def calculate_salary_expense(employee_list):
    """Calculate current, revised and additional salary."""

    if not employee_list:
        return {
            "current_total": 0,
            "revised_total": 0,
            "additional_expense": 0
        }

    # Calculate revised salaries.
    revised_employees = calculate_revised_salaries(employee_list)

    # Use reduce() to calculate current total salary.
    current_total = reduce(lambda accumulator, employee: accumulator + employee["salary"], employee_list, 0)

    # Use reduce() to calculate revised total salary.
    revised_total = reduce(lambda accumulator, employee: accumulator + employee["revised_salary"], revised_employees, 0 )

    # Calculate the additional salary expense.
    additional_expense = revised_total - current_total

    return {
        "current_total": current_total,
        "revised_total": revised_total,
        "additional_expense": additional_expense
    }


# ============================================================
# PART 7: FIND HIGHEST-PAID EMPLOYEE USING reduce()
# ============================================================

def get_highest_paid(employee_list):
    """Find the employee with the highest salary."""

    if not employee_list:
        return None

    # reduce() compares two employees at a time.
    # The employee with the larger salary is kept.
    highest = reduce(lambda employee1, employee2: employee1 if employee1["salary"] >= employee2["salary"] else employee2, employee_list)

    return highest


# ============================================================
# PART 8: SORT EMPLOYEES
# ============================================================

def sort_employees(employee_list, choice):
    """Sort employees using sorted() and lambda."""

    if choice == "1":
        # Sort by name
        return sorted(employee_list, key=lambda employee: employee["name"])

    elif choice == "2":
        # Sort by salary
        return sorted(employee_list, key=lambda employee: employee["salary"], reverse=True)

    elif choice == "3":
        # Sort by performance score
        return sorted(employee_list, key=lambda employee: employee["performance_score"], reverse=True)

    elif choice == "4":
        # Sort by experience
        return sorted(employee_list, key=lambda employee: employee["experience"], reverse=True)

    else:
        return None


# ============================================================
# PART 9: CLOSURE-BASED BONUS CALCULATOR
# ============================================================

def create_bonus_calculator(bonus_percentage):
    """
    Outer function accepts bonus percentage.

    The inner function remembers bonus_percentage.
    This is a closure.
    """

    def calculate_bonus(salary):
        # The inner function can access bonus_percentage
        # from the enclosing scope.
        return salary * bonus_percentage / 100

    # Return the inner function.
    return calculate_bonus


# Create separate bonus calculator functions.
bonus_5 = create_bonus_calculator(5)
bonus_10 = create_bonus_calculator(10)
bonus_15 = create_bonus_calculator(15)


# ============================================================
# PART 10 AND 11:
# NESTED DEPARTMENT REPORT + LEGB DEMONSTRATION
# ============================================================

def generate_department_report(employee_list, department):

    # --------------------------------------------------------
    # ENCLOSING SCOPE
    # --------------------------------------------------------
    # 'department' belongs to generate_department_report().
    # Therefore it is an enclosing variable for the
    # nested functions below.

    # --------------------------------------------------------
    # LOCAL SCOPE
    # --------------------------------------------------------
    report_title = "Department Performance Report"

    # --------------------------------------------------------
    # NESTED FUNCTION 1
    # --------------------------------------------------------

    def employee_count():
        return len(employee_list)

    # --------------------------------------------------------
    # NESTED FUNCTION 2
    # --------------------------------------------------------

    def total_salary():
        return reduce(lambda accumulator, employee: accumulator + employee["salary"], employee_list, 0)

    # --------------------------------------------------------
    # NESTED FUNCTION 3
    # --------------------------------------------------------

    def average_salary():
        count = employee_count()

        if count == 0:
            return 0

        return total_salary() / count

    # --------------------------------------------------------
    # NESTED FUNCTION 4
    # --------------------------------------------------------

    def average_performance():
        count = employee_count()

        if count == 0:
            return 0

        total_score = reduce(lambda accumulator, employee: accumulator + employee["performance_score"], employee_list, 0)

        return total_score / count

    # --------------------------------------------------------
    # LEGB DEMONSTRATION
    # --------------------------------------------------------

    # company_name is not local to this function.
    # Python checks:
    #
    # L = Local
    # E = Enclosing
    # G = Global  -> company_name is found here
    # B = Built-in
    #
    # Therefore Python uses the global company_name.
    current_company = company_name

    # 'department' is found in the enclosing scope.
    current_department = department

    # 'report_title' is a local variable of this function.
    current_title = report_title

    # Return the complete report as a dictionary.
    return {
        "company": current_company,
        "report_title": current_title,
        "department": current_department,
        "employee_count": employee_count(),
        "total_salary": total_salary(),
        "average_salary": average_salary(),
        "average_performance": average_performance()
    }


# ============================================================
# DISPLAY DEPARTMENT REPORT
# ============================================================

def display_department_report(employee_list, department):

    department_employees = get_by_department(employee_list, department)

    # Handle department not found.
    if not department_employees:
        print("\nDepartment not found.")
        return

    report = generate_department_report(department_employees, department)

    print("\n========== DEPARTMENT REPORT ==========")
    print("Company              :", report["company"])
    print("Report Title         :", report["report_title"])
    print("Department           :", report["department"])
    print("Number of Employees  :", report["employee_count"])
    print("Total Salary         :", report["total_salary"])
    print("Average Salary      :", round(report["average_salary"], 2))
    print("Average Performance  :", round(report["average_performance"], 2))

# ============================================================
# DISPLAY BONUS EXAMPLE
# ============================================================

def display_bonus_calculators():
    """Demonstrate the closure-based bonus calculators."""

    salary = 60000

    print("\n========== BONUS CALCULATORS ==========")
    print("Salary:", salary)
    print("5% Bonus :", bonus_5(salary))
    print("10% Bonus:", bonus_10(salary))
    print("15% Bonus:", bonus_15(salary))


# ============================================================
# MENU OPTION 5: DISPLAY REVISED SALARIES
# ============================================================

def display_revised_salaries(employee_list):

    if not employee_list:
        print("\nNo employees available.")
        return

    revised = calculate_revised_salaries(employee_list)

    print("\n========== REVISED SALARIES ==========")

    for employee in revised:
        print(
            employee["name"],
            "-> Current:",
            employee["salary"],
            "| Revised:",
            employee["revised_salary"]
        )

# ============================================================
# MENU OPTION 6: DISPLAY SALARY EXPENSE
# ============================================================

def display_salary_expense(employee_list):

    if not employee_list:
        print("\nNo employees available.")
        return

    expense = calculate_salary_expense(employee_list)

    print("\n========== SALARY EXPENSE ==========")
    print("Current Total Salary    :", expense["current_total"])
    print("Revised Total Salary    :", expense["revised_total"])
    print("Additional Salary Cost  :", expense["additional_expense"])


# ============================================================
# MENU OPTION 7: DISPLAY HIGHEST PAID EMPLOYEE
# ============================================================

def display_highest_paid(employee_list):

    employee = get_highest_paid(employee_list)

    if employee is None:
        print("\nNo employees available.")
        return

    print("\n========== HIGHEST-PAID EMPLOYEE ==========")
    print("Name       :", employee["name"])
    print("Department :", employee["department"])
    print("Salary     :", employee["salary"])


# ============================================================
# MENU OPTION 8: SORT EMPLOYEES
# ============================================================

def display_sorted_employees(employee_list):

    print("\nSort employees by:")
    print("1. Name")
    print("2. Salary")
    print("3. Performance score")
    print("4. Experience")

    choice = input("Enter your choice: ")

    sorted_list = sort_employees(employee_list, choice)

    if sorted_list is None:
        print("Invalid sorting choice.")
        return

    print("\n========== SORTED EMPLOYEES ==========")

    for employee in sorted_list:
        print(
            employee["name"],
            "| Salary:",
            employee["salary"],
            "| Performance:",
            employee["performance_score"],
            "| Experience:",
            employee["experience"]
        )


# ============================================================
# MENU OPTION 4: SEARCH DEPARTMENT
# ============================================================

def search_department(employee_list):

    department = input("Enter department name: ")

    result = get_by_department(
        employee_list,
        department
    )

    if not result:
        print("\nDepartment does not exist.")
        return

    display_employees(result)


# ============================================================
# VALIDATE ALL EMPLOYEES
# ============================================================

def get_valid_employees(employee_list):

    valid_employees = []

    for employee in employee_list:

        if validate_employee(employee):
            valid_employees.append(employee)

        else:
            print(
                "Invalid employee data found for:",
                employee.get("name", "Unknown")
            )

    return valid_employees


# ============================================================
# MAIN MENU
# ============================================================

def main():

    # Validate the employee data before starting.
    valid_employees = get_valid_employees(employees)

    # Continue showing the menu until the user chooses Exit.
    while True:

        print("\n")
        print("==============================================")
        print(" EMPLOYEE PERFORMANCE ANALYSIS SYSTEM")
        print("==============================================")
        print("1. Display all employees")
        print("2. Display employee names")
        print("3. Display high performers")
        print("4. Search by department")
        print("5. Calculate revised salaries")
        print("6. Calculate total salary expense")
        print("7. Find highest-paid employee")
        print("8. Sort employees")
        print("9. Generate department report")
        print("10. Exit")
        print("==============================================")

        choice = input("Enter your choice: ")

        # ----------------------------------------------------
        # OPTION 1
        # ----------------------------------------------------

        if choice == "1":
            display_employees(valid_employees)

        # ----------------------------------------------------
        # OPTION 2
        # ----------------------------------------------------

        elif choice == "2":
            names = get_employee_names(valid_employees)

            print("\nEmployee Names:")
            print(names)

        # ----------------------------------------------------
        # OPTION 3
        # ----------------------------------------------------

        elif choice == "3":
            high_performers = get_high_performers(
                valid_employees
            )

            display_employees(high_performers)

        # ----------------------------------------------------
        # OPTION 4
        # ----------------------------------------------------

        elif choice == "4":
            search_department(valid_employees)

        # ----------------------------------------------------
        # OPTION 5
        # ----------------------------------------------------

        elif choice == "5":
            display_revised_salaries(valid_employees)

        # ----------------------------------------------------
        # OPTION 6
        # ----------------------------------------------------

        elif choice == "6":
            display_salary_expense(valid_employees)

        # ----------------------------------------------------
        # OPTION 7
        # ----------------------------------------------------

        elif choice == "7":
            display_highest_paid(valid_employees)

        # ----------------------------------------------------
        # OPTION 8
        # ----------------------------------------------------

        elif choice == "8":
            display_sorted_employees(valid_employees)

        # ----------------------------------------------------
        # OPTION 9
        # ----------------------------------------------------

        elif choice == "9":
            department = input(
                "Enter department name: "
            )

            display_department_report(
                valid_employees,
                department
            )

        # ----------------------------------------------------
        # OPTION 10
        # ----------------------------------------------------

        elif choice == "10":
            print("\nThank you for using the system.")
            break

        # ----------------------------------------------------
        # INVALID MENU CHOICE
        # ----------------------------------------------------

        else:
            print("\nInvalid menu choice. Please try again.")


# ============================================================
# START THE PROGRAM
# ============================================================

if __name__ == "__main__":
    main()

# Sample Execution 1

# Input

# Enter your choice: 2

# Output

#
# Employee Names:
# ['Ravi', 'Anitha', 'Kiran', 'Meena', 'Arjun']

#---
# Sample Execution 2

# Input

# Enter your choice: 3

# Output

# ========== EMPLOYEE DETAILS ==========
# --------------------------------------
# ID                : 101
# Name              : Ravi
# Department        : IT
# Salary            : 60000
# Performance Score : 88
# Experience        : 5 years
# --------------------------------------
# ID                : 103
# Name              : Kiran
# Department        : IT
# Salary            : 75000
# Performance Score : 94
# Experience        : 7 years
# --------------------------------------
# ID                : 104
# Name              : Meena
# Department        : Finance
# Salary            : 55000
# Performance Score : 81
# Experience        : 6 years
# --------------------------------------

# These three employees have performance scores of **80 or above**.

# ---

# # Sample Execution 3

# ### Input

# Enter your choice: 9
# Enter department name: IT

# ### Output

# ========== DEPARTMENT REPORT ==========
# Company              : Edukron
# Report Title         : Department Performance Report
# Department           : IT
# Number of Employees  : 3
# Total Salary         : 175000
# Average Salary       : 58333.33
# Average Performance  : 82.33

# # Important results

# For all five employees:

# Current total salary   = 275000
# Revised total salary   = 299250
# Additional salary cost = 24250

# The highest-paid employee is:

# Kiran
# Salary: 75000
```

```python
# # Concepts Used

# ## 1. Functions

# A function is a reusable block of code.

# Example:

# def display_employees(employee_list):
#     ...

# Instead of writing the employee-display code repeatedly, we can call:

# display_employees(employees)

# ---

# ## 2. Lambda Expressions

# A lambda is a small anonymous function.

# Example:

# lambda employee: employee["name"]

# This gets the name from an employee dictionary.

# Another example:

# lambda employee: employee["salary"]

# This gets the salary.

# ---

# ## 3. `map()`

# `map()` applies a function to every item.

# In this project:

# names = list(
#     map(lambda employee: employee["name"], employee_list))

# For example:

# Employees
#    |
# map()
#    |
# Ravi, Anitha, Kiran, Meena, Arjun


# ---

# ## 4. `filter()`

# `filter()` selects only items that satisfy a condition.

# Example:

# filter( lambda employee: employee["performance_score"] >= 80, employee_list)

# Only employees with a score of 80 or more are selected.

# ---

# ## 5. `reduce()`

# `reduce()` combines multiple values into one final value.

# Example:

# reduce(lambda accumulator, employee: accumulator + employee["salary"], employee_list, 0)

# It calculates the total salary.

# Conceptually:

# 0 + 60000
# 60000 + 45000
# 105000 + 75000
# 180000 + 55000
# 235000 + 40000
# = 275000

# ---

# ## 6. Nested Functions

# A nested function is a function defined inside another function.

# Example:

# def generate_department_report(employee_list, department):

#     def employee_count():
#         return len(employee_list)

#     def total_salary():
#         ...

#     def average_salary():
#         ...

# The smaller functions are inside `generate_department_report()`.

# ---

# ## 7. Closure

# A closure occurs when an inner function remembers a value from its enclosing function.

# Example:

# def create_bonus_calculator(bonus_percentage):

#     def calculate_bonus(salary):
#         return salary * bonus_percentage / 100

#     return calculate_bonus

# Then:

# bonus_10 = create_bonus_calculator(10)

# `bonus_10` remembers that the bonus percentage is `10`.

# Therefore:

# bonus_10(60000)

# returns:

# 6000

# ---

# ## 8. Local Scope

# A local variable exists inside a function.

# Example:

# def generate_department_report(employee_list, department):

#     report_title = "Department Performance Report"

# `report_title` is local to `generate_department_report()`.

# ---

# ## 9. Enclosing Scope

# An enclosing variable belongs to an outer function and can be accessed by its nested functions.

# Example:

# def generate_department_report(employee_list, department):

#     # department belongs to the outer function

#     def employee_count():
#         print(department)

# The nested function can access `department` from the enclosing function.

# ---

# ## 10. Global Scope

# A global variable is created outside functions.

# In this project:

# company_name = "Edukron"

# It can be accessed by functions.

# Inside the report function:

# current_company = company_name

# Python eventually finds `company_name` in the global scope.

# ---

# ## 11. Built-in Scope

# Built-in scope contains Python's built-in names and functions.

# Examples:

# len()
# print()
# list()
# int()
# str()

# For example:

# len(employee_list)

# uses the built-in `len()` function.

# ---

# # LEGB Order

# Python searches for names in this order:
# L → Local
# E → Enclosing
# G → Global
# B → Built-in

# For example, when Python sees:

# print(value)

# it searches:

# 1. Local scope
#        |
# 2. Enclosing scope
#        |
# 3. Global scope
#        |
# 4. Built-in scope

# Python uses the **first matching name** it finds.

# ---

# # Normal Function vs Lambda

# | Normal Function                   | Lambda                                          |
# | --------------------------------- | ----------------------------------------------- |
# | Uses `def`                        | Uses `lambda`                                   |
# | Can contain multiple statements   | Usually one expression                          |
# | Can have a descriptive name       | Usually anonymous                               |
# | Better for larger logic           | Good for short operations                       |
# | Easier to reuse for complex logic | Convenient with `map()`, `filter()`, `sorted()` |

# Example normal function:

# def square(number):
#     return number ** 2

# Equivalent lambda:

# lambda number: number ** 2

# For simple operations, lambda is convenient.

# For complicated logic, a normal function is usually easier to read.

# ---

# # Loops vs Functional Programming Tools

# ### Using a loop

# For example, extracting names:

# names = []

# for employee in employees:
#     names.append(employee["name"])

# ### Using `map()`

# names = list(map(lambda employee: employee["name"], employees))

# Both approaches can produce the same result.

# ### Functional programming tools

# map()     → transform data
# filter()  → select data
# reduce()  → combine data
# sorted()  → sort data
# lambda    → create short functions

# ### Simple comparison

# | Loops                                      | Functional Tools                                       |
# | ------------------------------------------ | ------------------------------------------------------ |
# | Explicit and beginner-friendly             | More compact                                           |
# | Good for complex logic                     | Good for transformations                               |
# | Uses `for`/`while`                         | Uses `map()`, `filter()`, `reduce()`                   |
# | Often easier to debug step-by-step         | Can be concise                                         |
# | Can modify collections if written that way | Encourages processing data without direct modification |

# In this project, both ordinary functions and functional tools are used together so that each tool performs the job it is best suited for.
```
