# Part 7: Profit Margin Analysis

```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([

    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([

    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

profit_data = sales_data - cost_data

print("Profit Matrix:")
print(profit_data)
```

```python
# Question 35: Profit Margin Matrix
# Calculate the profit margin percentage for every store-category combination.

# Use:

# profit_margin = (profit_data / sales_data) * 100
# Round the result to two decimal places.


profit_data = sales_data - cost_data

profit_margin = np.round((profit_data / sales_data) * 100, 2)

print(profit_margin)
```

```python
# Question 36: Store-Wise Profit Margin
# Calculate the overall profit margin percentage for each store using:

# Store Profit Margin =
# Total Store Profit / Total Store Sales × 100

store_profit = profit_data.sum(axis=1)
store_sales = sales_data.sum(axis=1)

store_profit_margin = np.round((store_profit / store_sales) * 100, 2)

print(store_profit_margin)
```

```python
# Question 37: Category-Wise Profit Margin
# Calculate the overall profit margin percentage for each product category.

category_profit = profit_data.sum(axis=0)
category_sales = sales_data.sum(axis=0)

category_profit_margin = np.round((category_profit / category_sales) * 100, 2)

print(category_profit_margin)
```

```python
# Question 38: Best Store by Profit Margin
# Identify the store with the highest overall profit margin.

best_store_index = np.argmax(store_profit_margin)
best_store = stores[best_store_index]

print(best_store)
```

```python
# Question 39: Best Category by Profit Margin
# Identify the product category with the highest profit margin.

best_category_index = np.argmax(category_profit_margin)
best_category = categories[best_category_index]

print(best_category)
```

```python
# Question 40: Low-Margin Transactions
# Management considers a profit margin below 22% to be weak.

# Find all store-category combinations where the profit margin is below 22%.

# Display:

# Store
# Category
# Sales
# Cost
# Profit
# Profit margin
# Use Boolean indexing.

# Calculate profit margin
# profit_margin = (profit_data / sales_data) * 100

profit_margin = (profit_data / sales_data) * 100

low_margin_mask = profit_margin < 22

store_idx, category_idx = np.where(low_margin_mask)

if len(store_idx) == 0:
    print("No store-category combinations found with profit margin below 22%.")
else:
  for i, j in zip(store_idx, category_idx):
   store_name = stores[i]
   category_name = categories[j]
   sales = sales_data[i,j]
   cost = cost_data[i, j]
   profit = profit_data[i, j]
   profit_margin = round(profit_margin[i, j],2)

   print(f"\nStore: {store_name}")
   print(f"Category: {category_name}")
   print(f"  Sales: {sales}")
   print(f"  Cost: {cost}")
   print(f"  Profit: {profit}")
   print(f"  Profit Margin: {profit_margin:.2f}%")
```
