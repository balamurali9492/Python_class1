```python
#Find the sum of square roots of all numbers in a list.
Num=[1,5,7,8]
sum=0
for i in Num:
  sum=sum+math.sqrt(i)
print(sum)
```

```python
#Calculate compound interest using math.pow().
#p r n t
#A=P(1+nr​)nt
Principle = 5000    #(p)
interest = 5 / 100  #(r)
time = 2            #(n)
compound = 4        #(t)

A = Principle * math.pow((1 + interest/compound), compound*time)

Compoundinterest = A - 10000

print(A)
print(Compoundinterest)
```

```python
#Find the area and circumference of a circle using math.pi.
r=5
area = math.pi * r * r
circumference = 2 * math.pi * r
print(area)
print(circumference)
```

```python
#Create a scientific calculator that supports square root, factorial, logarithm, power, trigonometric functions, GCD, and LCM.
print("1. Square Root")
print("2. Factorial")
print("3. Logarithm")
print("4. Power")
print("5. Trigonometry")
print("6. GCD")
print("7. LCM")

choice = int(input("Enter choice: "))

if choice == 1:
    n = int(input("Enter number: "))
    print(math.sqrt(n))

elif choice == 2:
    n = int(input("Enter number: "))
    print(math.factorial(n))

elif choice == 3:
    n = float(input("Enter number: "))
    print(math.log(n))

elif choice == 4:
    a = int(input("Enter base: "))
    b = int(input("Enter power: "))
    print(math.pow(a, b))

elif choice == 5:
    angle = int(input("Enter angle: "))
    print("Sin =", math.sin(math.radians(angle)))
    print("Cos =", math.cos(math.radians(angle)))
    print("Tan =", math.tan(math.radians(angle)))

elif choice == 6:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print(math.gcd(a, b))

elif choice == 7:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print((a * b) // math.gcd(a, b))

else:
    print("Invalid choice")
```
